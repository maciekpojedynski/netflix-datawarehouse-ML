## 3. Cleaning and Imputation of Recommendation Score

```python
print(df_silver['recommendation_score'])

median_value = df_silver['recommendation_score'].median()
df_silver['is_recommendation_score_missing'] = df_silver['recommendation_score'].isna().astype(int)
df_silver['recommendation_score'] = df_silver['recommendation_score'].fillna(median_value)
df_silver.info()

#Verification
print(df_silver['recommendation_score'].isnull().sum())
```

## 4. Cleaning and Imputation of algorithm Version

```python
print(df_silver['algorithm_version'])
print(df_silver['algorithm_version'].unique())
print(df_silver['algorithm_version'].value_counts())
#SQL WINDOW FUNCTION TO FILL MISSING VALUES
df_silver['is_algorithm_version_missing'] = df_silver['algorithm_version'].isna().astype(int)
df_silver['algorithm_version'] = df_silver['algorithm_version'].fillna('TBD')
```

## 7. Final Output: Saving Silver Layer to CSV

```python
output_file = 'netflix_silver_layer_logs1.csv'

# Save the cleaned df_silverFrame. index=False is crucial for TSQL BULK INSERT.
df_silver.to_csv(
    output_file,
    sep = ',',
    index=False # Ensure this is set to False for the final output!
)
```
